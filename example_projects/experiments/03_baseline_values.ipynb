# Just the values: an experiment with no scenarios

Not every "experiment" is a comparison. Sometimes you just want the pipeline's
plots and summaries for one run — a static snapshot, not a drift check. The
same `ExperimentDef` handles this: leave `scenarios` empty (or supply exactly
one, if you want it named) and nothing about the schema changes; `comparison`
and `summaries` simply have nothing to compare against.

```python
import polars as pl
from decider.experiments import Experiment, OutcomeCounts, RunStatus
from example_projects.experiments.credit_pipeline import build

ENTRY = "example_projects.experiments.credit_pipeline:build"
pipeline = build()
df = pl.DataFrame({
    "income": [1000.0, 2000.0, 1500.0, 800.0],
    "debt": [350.0, 900.0, 700.0, 300.0],
})
df
```

```python
exp = Experiment("snapshot", pipeline, df, entry=ENTRY)
exp.plot(OutcomeCounts(type="outcome_counts", column="approved"))

result = exp.run()
assert result.status is RunStatus.COMPLETED
result.status, len(result.scenarios)
```

No scenarios ran, but the baseline still ran once, and the plot still has
something to show — the point of "static value" experiments: get values out,
with no second pipeline to diff against.
