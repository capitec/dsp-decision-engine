# Just the values: an experiment with no scenarios

Not every experiment is a comparison. With no scenarios only the baseline
runs, and the metrics describe it: here, approvals from the flow's output, and
the distribution of `ratio` — an intermediate value, read from the step that
writes it with `nodes=`.

```python
import polars as pl
from decider.experiments import (Bars, Changes, Describe, Divergence, Experiment, Progress, RunStatus,
                                 ValueCounts)
from example_projects.experiments.credit_pipeline import build

ENTRY = "example_projects.experiments.credit_pipeline:build"
pipeline = build()
df = pl.DataFrame({
    "income": [1000.0, 2000.0, 1500.0, 800.0],
    "debt": [250.0, 500.0, 700.0, 250.0],
})
df
```

```python
result = (Experiment("snapshot", pipeline, df, entry=ENTRY)
          .metric(ValueCounts(column="approved"), Describe(column="ratio", nodes="credit/ratio"))
          .plot(Bars(metric="value_counts"))
          .run())
result.summaries["value_counts"]["baseline"], result.summaries["describe"]["baseline"]
```
